# ETAPA: Imports e paths


```python
from __future__ import annotations

import json
from pathlib import Path
from datetime import datetime

import pandas as pd

pd.set_option('display.max_columns', 250)
pd.set_option('display.width', 220)

BASE_DIR = Path('/home/wilson/Maringa/fase_1_diagnostico')
OUT_R1 = Path('/home/wilson/Maringa/fase_1_diagnostico/3ª-Retomada/outputs/R1')
OUT_R2 = Path('/home/wilson/Maringa/fase_1_diagnostico/3ª-Retomada/outputs/R2')
OUT_R3 = Path('/home/wilson/Maringa/fase_1_diagnostico/3ª-Retomada/outputs/R3')
ANEXOS = OUT_R3 / 'anexos'
OUT_R3.mkdir(parents=True, exist_ok=True)
ANEXOS.mkdir(parents=True, exist_ok=True)

RUN_TS = datetime.now().strftime('%Y-%m-%d_%H%M%S')

p_r1_sum = OUT_R1 / 'r1_summary.json'
p_r1_rep = OUT_R1 / 'r1_report.md'

p_r2_sum = OUT_R2 / 'r2_summary.json'
p_r2_domain = OUT_R2 / 'r2_domain_summary.csv'
p_r2_cov = OUT_R2 / 'r2_date_coverage.csv'
p_r2_forno = OUT_R2 / 'r2_forno_summary.csv'
p_r2_parse = OUT_R2 / 'r2_parse_audit.csv'

required = [p_r1_sum, p_r1_rep, p_r2_sum, p_r2_domain, p_r2_cov, p_r2_forno, p_r2_parse]
for p in required:
    if not p.exists():
        raise FileNotFoundError(f'Arquivo requerido ausente: {p}')

print('RUN_TS:', RUN_TS)
print('BASE_DIR:', BASE_DIR)
print('OUT_R3:', OUT_R3)

```

# ETAPA: Carregar evidencias R1 e R2 (finais)


```python
r1 = json.loads(p_r1_sum.read_text(encoding='utf-8'))
r2 = json.loads(p_r2_sum.read_text(encoding='utf-8'))

df_domain = pd.read_csv(p_r2_domain)
df_cov = pd.read_csv(p_r2_cov)
df_forno = pd.read_csv(p_r2_forno)
df_parse = pd.read_csv(p_r2_parse)

print('R1 summary:')
print(json.dumps(r1, ensure_ascii=False, indent=2))
print('\nR2 summary:')
print(json.dumps(r2, ensure_ascii=False, indent=2))

print('\nDomain summary (head 20):')
display(df_domain.head(20))

print('\nDate coverage (head 20):')
display(df_cov.head(20))

print('\nForno summary (head 20):')
display(df_forno.head(20))

```

# ETAPA: Localizar arquivos conceituais (PROPNT anterior) por nome de arquivo


```python
CONCEPT_FILES = [
    'RESUMO DO PROJETO.md',
    'DISCUSSÃO SOBRE QUALIDADE DE DADOS.md',
    'DISCUSSÃO SOBRE QUANTIDADE DE DADOS.md',
    'Quantidade de Dados - uma visão preliminar.md'
]


def find_files_by_name(base: Path, filename: str) -> list[Path]:
    cands = []
    for p in base.rglob(filename):
        if not p.is_file():
            continue
        # ignorar copias congeladas em ANEXOS para evitar ambiguidade
        try:
            if p.is_relative_to(ANEXOS):
                continue
        except Exception:
            pass
        cands.append(p)
    return sorted(cands)


def choose_unique(cands: list[Path], label: str) -> Path:
    if len(cands) == 1:
        return cands[0]
    print(f'ERRO: arquivo conceitual nao-unico ou nao-encontrado: {label}')
    print('Candidatos:')
    for i, p in enumerate(cands, 1):
        print(f'  {i:02d}: {p}')
    raise RuntimeError(f'Ambiguidade ou ausencia: {label}')

concept_paths = {}
for fn in CONCEPT_FILES:
    cands = find_files_by_name(BASE_DIR, fn)
    concept_paths[fn] = choose_unique(cands, fn)

print('Arquivos conceituais localizados:')
for k, v in concept_paths.items():
    print('-', k, '->', v)

# Ler conteudo integral
concept_text = {}
for fn, p in concept_paths.items():
    concept_text[fn] = p.read_text(encoding='utf-8', errors='strict')

print('Leitura concluida dos arquivos conceituais.')

```

# ETAPA: Copiar arquivos conceituais para anexos (freeze local) e gerar Relatorio RAW (R3)


```python
relatorio_path = Path('/home/wilson/Maringa/fase_1_diagnostico/3ª-Retomada/outputs/R3/Relatorio_RAW_Entregavel_1.md')

# Copiar (freeze) para anexos
anexos_written = []
for fn, src in concept_paths.items():
    dst = ANEXOS / fn
    dst.write_text(concept_text[fn], encoding='utf-8')
    anexos_written.append(dst)

# Numeros finais (R1/R2)
n_total = int(r2.get('n_total_files_manifest', 0))
n_csv = int(r2.get('n_csv_files', 0))
n_ok = int(r2.get('n_csv_ok', 0))
n_fail = int(r2.get('n_csv_fail', 0))
n_dom = int(r2.get('n_dominios', 0))

r1_total = int(r1.get('total_linhas_manifesto', 0))
r1_found = int(r1.get('encontrados_no_fs', 0))
r1_missing = int(r1.get('ausentes_no_fs', 0))
r1_dups = int(r1.get('duplicados_no_manifesto_linhas', 0))
r1_orphans = int(r1.get('orfaos_no_fs', 0))

lines = []
lines.append('# Relatorio RAW - Entregavel 1')
lines.append('')
lines.append(f'Geracao: {RUN_TS}')
lines.append('')

# Corpo conceitual: inserir como secoes, sem reescrita livre
# Estrategia: referenciar anexos e inserir o texto integral abaixo de cada titulo

lines.append('## Resumo do projeto')
lines.append('')
lines.append(concept_text['RESUMO DO PROJETO.md'])
lines.append('')

lines.append('## Discussao sobre qualidade de dados')
lines.append('')
lines.append(concept_text['DISCUSSÃO SOBRE QUALIDADE DE DADOS.md'])
lines.append('')

lines.append('## Discussao sobre quantidade de dados')
lines.append('')
lines.append(concept_text['DISCUSSÃO SOBRE QUANTIDADE DE DADOS.md'])
lines.append('')

lines.append('## Quantidade de dados - uma visao preliminar')
lines.append('')
lines.append(concept_text['Quantidade de Dados - uma visão preliminar.md'])
lines.append('')

# Evidencias objetivas (R1/R2)
lines.append('## Evidencias R1 - Coerencia catalogo versus filesystem')
lines.append('')
lines.append(
    f'R1 confirma coerencia total entre manifesto e filesystem. Universo catalogado: {r1_total} linhas. '
    f'Encontrados no filesystem: {r1_found}. Ausentes: {r1_missing}. '
    f'Duplicidades no manifesto (linhas): {r1_dups}. Orfaos no filesystem: {r1_orphans}.'
)
lines.append('')

lines.append('## Evidencias R2 - Diagnostico de leitura e estrutura (nivel diagnostico)')
lines.append('')
lines.append(
    f'O universo catalogado contem {n_total} arquivos, sendo {n_csv} CSV e {n_total - n_csv} nao-CSV. '
    f'Parsing dos CSV: ok={n_ok}, fail={n_fail}. Dominios identificados: {n_dom}.'
)
lines.append('')

lines.append('### Sumario por dominio (R2)')
lines.append('')
lines.append(df_domain.to_markdown(index=False))
lines.append('')

lines.append('### Sumario por forno (R2)')
lines.append('')
lines.append(df_forno.to_markdown(index=False))
lines.append('')

lines.append('### Cobertura temporal automaticamente inferida (R2)')
lines.append('')
if df_cov.shape[0] == 0:
    lines.append('Nao ha cobertura temporal consolidada no artefato r2_date_coverage.csv.')
else:
    lines.append(df_cov.to_markdown(index=False))
lines.append('')

# Referencias (caminhos locais) e anexos congelados
lines.append('## Evidencias e anexos (caminhos locais)')
lines.append('')
lines.append(f'- R1 summary: {p_r1_sum}')
lines.append(f'- R1 report: {p_r1_rep}')
lines.append(f'- R2 summary: {p_r2_sum}')
lines.append(f'- R2 domain summary: {p_r2_domain}')
lines.append(f'- R2 date coverage: {p_r2_cov}')
lines.append(f'- R2 forno summary: {p_r2_forno}')
lines.append(f'- R2 parse audit: {p_r2_parse}')
lines.append('')
lines.append('- Anexos congelados (copias locais dos textos conceituais):')
for p in anexos_written:
    lines.append(f'  - {p}')
lines.append('')

relatorio_path.write_text('\n'.join(lines), encoding='utf-8')
print('Relatorio salvo em:', relatorio_path)
print('Anexos gravados em:', ANEXOS)

```
